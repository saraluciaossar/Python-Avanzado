## 1. Calculadora de l'índex de massa corporal

Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en: https://muysalud.com/salud/indice-de-masa-corporal-imc-que-es-y-como-se-calcula/ Índice de masa corporal IMC que es y como se calcula.

La funció ha de classificar el resultat en les seves respectives categories.

Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduïdes per l’usuari/a no estan en el format adequat o no pren valors raonables.


```python
def IMC():
    """
    Esta función le permite al usuario/a calcular su IMC. Para ello pide peso en KG y altura en metros.
    Retorna el IMC total y el rango de IMC en que se encuentra la persona.
    """
    while True:
        try:
            peso = float(input("Ingrese su peso en Kg: "))
            if peso <= 1 or peso >= 350:
                print("Ingrese un peso entre 1 y 350 kg")
                continue
            break
        except ValueError:
            print("Dato incorrecto, por favor inténtelo de nuevo")

    while True:
        try:
            altura = float(input("Ingrese su altura en metros: "))
            if altura <= 0 or altura >= 2.5:
                print("Ingrese una altura entre 0.1 y 2.5 m")
            break
        except ValueError:
            print("Dato incorrecto, por favor inténtelo de nuevo")
    
    IMC_total = round(peso / (altura ** 2), 2)
    
    if IMC_total < 18.5: 
        rango = "bajo peso"
    elif 18.5 <= IMC_total <= 24.9:
        rango = "peso normal"
    elif 24.9 < IMC_total <= 29.9:
        rango = "sobrepeso"
    else:
        rango = "obesidad"

    return IMC_total, rango
```

```python
IMC_total, rango = IMC()
print(f"Su IMC es: {IMC_total}")
print(f"Usted tiene: {rango}")
```

## 2. Convertidor de temperatures

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un print() no es pot guardar mai).

Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduides per l’usuari/a no estan en el format adequat.

(EXTRA): Pensa una manera d’emmagatzemar totes les possibles conversions en un sol objecte (Llista? Diccionari? DataFrame?) en comptes d’escriure molts if else en funció de la temperatura d’origen i la temperatura de destí.

```python
def calculadora_temperatura():
   """
   Esta función le permite al usuario/a convertir medidas de temperatura. 
   Valor: float
   Medida de entrada: Celsius (°C), Fahrenheit (°F) o Kelvin (K)
   Retorna: la temperatura ingresada y sus equivalencias en las otras dos unidades de medida.
   """
   cero_absoluto = {
       "celsius": -273.15,
       "fahrenheit": -459.67,
       "kelvin": 0
   }

   conversiones = {
       ("celsius", "fahrenheit"): lambda t: (t * 9/5) + 32,
       ("celsius", "kelvin"):     lambda t: t + 273.15,
       ("fahrenheit", "celsius"): lambda t: (t - 32) * 5/9,
       ("fahrenheit", "kelvin"):  lambda t: (t - 32) * 5/9 + 273.15,
       ("kelvin", "celsius"):     lambda t: t - 273.15,
       ("kelvin", "fahrenheit"):  lambda t: (t - 273.15) * 9/5 + 32,
   }

   while True:
       unidad = input("¿En qué unidad está la temperatura? (Celsius, Fahrenheit, Kelvin): ").strip().lower()
       if unidad not in cero_absoluto:
           print("Error en la temperatura ingresada, por favor compruebe que no haya error de typos")
           continue
       break

   while True:
       try:
           temperatura = float(input(f"Ingrese la temperatura en {unidad}: "))
           if temperatura < cero_absoluto[unidad]:
               print(f"Esta no es una temperatura correcta en {unidad}, por favor inténtelo de nuevo")
               continue
           break
       except ValueError:
           print("La temperatura debe ingresarse en números, por favor inténtelo de nuevo")

   resultado = {}
   for (unidad_origen, unidad_destino), formula in conversiones.items():
       if unidad_origen == unidad:
           resultado[unidad_destino] = round(formula(temperatura), 2)

   resultado[unidad] = temperatura  # añades la original al mismo diccionario

   return resultado
```

```python
resultado = calculadora_temperatura()
print(resultado)
```

## 3. Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.

(EXTRA): Quina és la llargada mitjana de les paraules del text que has escrit? “Hola com va?” hauria de retornar (4+3+2) / 3 = 3

```python
import re       # Modulo de patrones para eliminar : "re.sub(patrón, reemplazo, texto)"

def contador_palabras():
    """
    Esta función le permite al usuario/a calcular el número de palabras en un texto recibido y contar la media de palabras que tiene el texto. 
    Retorna: diccionario con todas las conversiones
    """
    while True:
        texto = input("Ingresa un texto: ")
        if not texto.strip():
            print("El texto no puede estar vacío, inténtalo de nuevo")
            continue
        if texto.strip().isnumeric():
            print("Debes introducir texto, no solo números, inténtalo de nuevo")
            continue
        break

    texto = texto.lower()                        
    texto = texto.strip()
    texto = re.sub(r"[^\w\s]", "", texto)   # Elimina todo lo que no sea letra o espacio
    texto = re.sub(r"\d+", "", texto)       # Eliminan números que se hayan quedado pegados con las palabras (ej: sara1991)

    dict_palabras = {}
    for palabra in texto.split():
        if palabra in dict_palabras:
            dict_palabras[palabra] += 1
        else:
            dict_palabras[palabra] = 1

    conteo_palabras = texto.split()
    total_letras = 0
    for palabra in conteo_palabras:
        total_letras += len(palabra)

    longitud_media = total_letras / len(conteo_palabras)

    return dict_palabras, longitud_media
```

```python
palabras, media = contador_palabras()
print(f"La cantidad de palabras es: {palabras}")
print(f"La longitud media es: {media} letras")
```

## 4. Diccionari invers (amb possibilitat de duplicats)

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.

```python
def diccionario_inverso(diccionario):
    """
    Esta función invierte la estructura de un diccionario de "clave : valor" a "valor : clave". 
    Además, si hay valores duplicados, genera una advertencia de los mismos y los enlista con su clave.
    Retorna el resultado del diccionario inverso.
    """    
    inverso = {}
    for clave, valor in diccionario.items():
        if valor in inverso:
            if isinstance(inverso[valor], list):
                inverso[valor].append(clave)
            else:
                inverso[valor] = [inverso[valor], clave]
        else:
            inverso[valor] = clave
    
    resultado = {}
    for clave, valor in inverso.items():
        if isinstance(valor, list):
            print(f"¡Advertencia! el elemento '{clave}' tiene valores duplicados: {valor}")
            resultado[clave] = valor
        else:
            resultado[clave] = valor
    
    return resultado
```

```python
# Sin duplicados
encuesta_1 = {"Lucia": 1991, "Rodrigo": 1986, "Sofia": 1996}
print(diccionario_inverso(encuesta_1))
```

```python
# Con duplicados
encuesta_2 = {"Victoria": 1989, "Nataly": 1989, "Alejandro": 1990}
print(diccionario_inverso(encuesta_2))
```

## 1. Comptador i endreçador de paraules d'un text.

El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. Fes servir l'arxiu "tu_me_quieres_blanca.txt".




```python
import re

def contador_palabras_archivo(archivo):
    """
    Esta función recibe la ruta de un archivo .txt
    y retorna la cantidad de palabras que contiene.
    """
    with open(archivo, "r", encoding="utf-8") as file:
        contenido = file.read().lower().strip()
        contenido = re.sub(r"[^\w\s]", "", contenido)
        contenido = re.sub(r"\d+", "", contenido)
        contenido = contenido.split()

    conteo_palabras = {}
    for palabra in contenido:
        conteo_palabras[palabra] = contenido.count(palabra)
    
    # Segmentado por letra con palabras
    segmentado = {}
    for palabra, frequencia in conteo_palabras.items():
        letra_inicial = palabra[0].upper()
        
        if letra_inicial not in segmentado:
            segmentado[letra_inicial] = {}
        
        segmentado[letra_inicial][palabra] = frequencia
    
    # Ordenado alfabéticamente
    resultado_ordenado = {}
    for letra in sorted(segmentado.keys()):
        resultado_ordenado[letra] = dict(sorted(segmentado[letra].items()))
    
    return resultado_ordenado
```

```python
texto_blanca = contador_palabras_archivo("tu_me_quieres_blanca.txt")
print(texto_blanca)
```

## 2. Conversió de tipus de dades.

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client:

[ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

```python
def conversion(data):
    """
    Esta función separa los elementos de una lista en dos: los que se pueden convertir a float y los que no.
    """
    convertidos = []
    no_convertidos = []

    for elemento in data:
        if isinstance(elemento, (list, tuple)):
            for item in elemento:
                try:
                    convertidos.append(float(item))
                except (ValueError, TypeError):
                    no_convertidos.append(item)
        else:
            try:
                convertidos.append(float(elemento))
            except (ValueError, TypeError):
                no_convertidos.append(elemento)

    return convertidos, no_convertidos
```

```python
lista_cliente = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]
conversion(lista_cliente)
```

## 1. Generador de Contrasenyes

Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

- longitud (int): Longitud de la contrasenya
- majuscules (bool = True): Si hi ha d’aparèixer majúscules
- minuscules (bool = True): Si hi ha d’aparèixer minúscules
- numeros (bool = True): Si hi ha d’aparèixer números
- signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)

Així doncs, si executem la funció de la següent manera:

crear_contrasenya(10, True, True, True, True)

Hauríem d’obtenir un output (que hem de poder guardar) de l’estil:

9Er,5Vn8P$

Assegura’t que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.

(EXTRA) Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l’ordinador (com si l’haguéssim seleccionat i fet ctrl+copy).

```python
import numpy as np
from numpy import random
import string

def crear_contraseña(longitud, mayusculas=True, minusculas=True, numeros=True, signos=False):
    """
    Esta función permite crear una contraseña aleatoria con la libreria de numpy
    """
    # Construir el pool de caracteres disponibles
    pool = ""
    if mayusculas: pool += string.ascii_uppercase        # A-Z
    if minusculas: pool += string.ascii_lowercase        # a-z
    if numeros:    pool += string.digits                 # 0-9
    if signos:     pool += ",.!@#$%^&*?-()/¡"            # especiales

    # Garantizar al menos un carácter de cada tipo activado
    obligatorios = []
    if mayusculas: obligatorios.append(random.choice(list(string.ascii_uppercase)))
    if minusculas: obligatorios.append(random.choice(list(string.ascii_lowercase)))
    if numeros:    obligatorios.append(random.choice(list(string.digits)))
    if signos:     obligatorios.append(random.choice(list(",.!@#$%^&*?-()/¡")))

    # Rellenar el resto hasta la longitud pedida
    resto = random.choice(list(pool), size=longitud - len(obligatorios))

    # Unir y mezclar todo
    todos = obligatorios + list(resto)
    random.shuffle(todos)

    # Unir la lista en un string
    contraseña = "".join(todos)
    return contraseña
```

## 2. Processament de dades simple

Una companya de feina ens ha demanat un favor, aprofitant que sap que estem aprenent a programar. Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. Necessita que processem les dades de manera automàtica, per extreure’n els resultats que necessita.

Utilitza el fitxer "historic_partits.txt"

Necessita un programa que retorni:

- El nombre total de gols que ha fet cada equip.
- El nom de l’equip més golejador.
- El nom de l’equip més golejat
- La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts)
- Com a referència, els equips amb més punts serien els següents:

imatge resultat partits
Pots utilitzar el format que et sigui més convenient.

```python
import pandas as pd

def analizar_partidos(fichero):
    """
    Esta función permite recibir un fichero de partidos en .txt para generar un resumen de la temporada de partidos arrojando:
    numero total de goles por cada equipo, el equipo mas goleador, el más goleado y la clasificación global.
    """
    # Cargar datos
    df = pd.read_csv(fichero, sep='\t', header=None)    # None para que no ponga header
    df.columns = ['equipo_local', 'marcador', 'equipo_visitante']
    df['goles_local'] = df['marcador'].str.split('-').str[0].astype(int)
    df['goles_visitante'] = df['marcador'].str.split('-').str[1].astype(int)

    # Preparar datos: Goles totales por equipo
    goles_local = df.groupby('equipo_local')['goles_local'].sum()
    goles_visitante = df.groupby('equipo_visitante')['goles_visitante'].sum()
    goles_totales = goles_local.add(goles_visitante, fill_value=0).astype(int)
    goles_totales.name = 'goles_a_favor'

    # Equipo más goleador
    mas_goleador = goles_totales.idxmax()

    # Equipo más goleado
    goles_recibidos_local = df.groupby('equipo_local')['goles_visitante'].sum()
    goles_recibidos_visitante = df.groupby('equipo_visitante')['goles_local'].sum()
    goles_recibidos = goles_recibidos_local.add(goles_recibidos_visitante, fill_value=0).astype(int)
    mas_goleado = goles_recibidos.idxmax()

    # Clasificación global
    puntos_clasificacion = {}
    for indice, fila in df.iterrows():
        local = fila['equipo_local']
        visitante = fila['equipo_visitante']
        gl = fila['goles_local']
        gv = fila['goles_visitante']

        puntos_clasificacion.setdefault(local, 0)
        puntos_clasificacion.setdefault(visitante, 0)

        if gl > gv:
            puntos_clasificacion[local] += 3
        elif gl < gv:
            puntos_clasificacion[visitante] += 3
        else:
            puntos_clasificacion[local] += 1
            puntos_clasificacion[visitante] += 1

    clasificacion = pd.DataFrame(puntos_clasificacion.items(), columns=['Equipo', 'Puntos'])
    clasificacion = clasificacion.sort_values('Puntos', ascending=False).reset_index(drop=True)
    clasificacion.index += 1

    # RESULTADOS
    df_goles = goles_totales.sort_values(ascending=False).reset_index()
    df_goles.columns = ['Equipo', 'Goles a favor']

    return {
        'mas_goleador': mas_goleador,
        'mas_goleado': mas_goleado,
        'goles_totales': df_goles,
        'clasificacion': clasificacion
    }
```
